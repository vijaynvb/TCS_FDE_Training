```python
import os
from anthropic import Anthropic
from dotenv import load_dotenv

load_dotenv()  # Load environment variables from .env file

client = Anthropic(
    api_key=os.environ.get("ANTHROPIC_API_KEY"),  # This is the default and can be omitted
)

message = client.messages.create(
    max_tokens=1024, # max tokens to be generated by the model 
    messages=[
        {
            "role": "user",
            "content": "Hello, Claude",
        }
    ],
    system="",
    model="claude-haiku-5-5",
)

print(message.content)
```

```python
# Tokenization

# systemmessage, humanmessage , aimessage 
# message [user, assistant]
# system

client.messages.count_tokens(
    model="claude-haiku-5-5",
    messages=[
        {"role": "user", "content": "Hello, world"},
        {"role": "assistant", "content": 'Hello! How can I help you today?'},
        {"role": "user", "content": "what is 2+2?"},
        {"role": "assistant", "content": '4'}
        ],
    system=[{"type": "text", "text": "you are it chat bot to help users to answer the issues related to networking only"}]
)
```

```python

message = client.messages.create(
    max_tokens=1024, # max tokens to be generated by the model 
    messages=[
        {
            "role": "user",
            "content": "Hello, my name is vijay",
        },
        {
            "role": "assistant",
            "content": "Hello!  vijay It's nice to meet you. How can I help you today?"
        },
        {
            "role": "user",
            "content": "what is 1+1",
        },
        {
            "role": "assistant",
            "content": "2"
        },
        {
            "role": "user",
            "content": "what is my name",
        }
    ], # context window
       # input 272k, 1M -> [message, systemprompt, tools, rag]
       # output 272k, 1M
       # cache memory in model, 

    model="claude-haiku-5-5",
)

print(message.content)
```
